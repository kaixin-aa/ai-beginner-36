# 文字表示与相似度实验
全部句子和主题向量为原创教学设定。词向量手工指定，没有下载或训练语义模型。

```python
from examples.text_core import *
import numpy as np
data = load_data()
print(np.__version__)
print(data['dimensions'])
```

```python
text = data['sentences'][0]
vector, tokens, ids = sentence_vector(text,data['vectors'])
print(text)
print(list(text))
print(tokens)
print(ids)
print(vector)
```

```python
rules,splits,alphabet = train_bpe(data['bpe_word_counts'])
for rule in rules: print(rule)
print(splits)
print('newlow',tokenize_bpe('newlow',rules,alphabet))
```

```python
chars,counts = character_vectors(data['sentences'])
encoded = [sentence_vector(text,data['vectors'])[0] for text in data['sentences']]
for i,j,case in data['pairs']:
    print(case,round(cosine(counts[i],counts[j]),6),round(cosine(encoded[i],encoded[j]),6))
```

```python
print(data['sentences'][6],data['sentences'][7])
print(cosine(encoded[6],encoded[7]))
assert cosine(encoded[6],encoded[7]) == 1
print('高分没有检查否定关系')
```

```python
print('尺度变大而方向不变',cosine(np.array([1,2]),np.array([10,20])))
print('正交',cosine([1,0],[0,1]))
print('相反',cosine([1,0],[-1,0]))
```

## 结果边界
字符分数考察重合。人工向量只保留预设主题，丢掉词序和否定信息。若换成训练好的语义模型，应单独核对模型版本、许可和实测，不能沿用本章数值。
