## 形状与索引
数组的第 0 轴对应学生，第 1 轴对应课程。

```python
print('shape', SCORES.shape)
print('ndim', SCORES.ndim)
print('size', SCORES.size)
print('小周的 AI 基础', SCORES[1, 2])
print('Python 列', SCORES[:, 0])
```

## 分别汇总学生和课程
axis 参数决定汇总哪个方向。

```python
result = summarize(SCORES)
print('学生均分', result['student_means'])
print('课程均分', result['subject_means'])
print('整体均分', result['overall_mean'])
print('批量与循环一致', np.allclose(result['student_means'], result['loop_means']))
```

```python
bonus = np.array([5, 0, 2])
print('调整成绩', np.minimum(SCORES + bonus, 100))
mask = (SCORES >= 60).all(axis=1)
print('全部及格', np.array(STUDENTS)[mask])
```
